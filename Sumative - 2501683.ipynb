````python
# Libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import time
import os


# Sorting Algorithms

def bubble_sort(arr):
    """Sorts a list in ascending order using Bubble Sort. Returns the sorted list."""
    arr = arr.copy()
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

def insertion_sort(arr):
    """Sorts a list in ascending order using Insertion Sort. Returns the sorted list."""
    arr = arr.copy()
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

def merge_sort(arr):
    """Sorts a list in ascending order using Merge Sort. Returns the sorted list."""
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

# Searching Algorithms ---

def linear_search(arr, target):
    """Searches for target in arr. Returns the index if found, -1 otherwise."""
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

def binary_search(arr, target):
    """Searches for target in a SORTED arr. Returns the index if found, -1 otherwise."""
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

# Datasets
url = "https://raw.githubusercontent.com/KennethNweke/HopeClass/refs/heads/main/student_performance.csv"
spdf = pd.read_csv(url)


print("\nSetup complete. All algorithms and datasets are ready.")
````

3. Has a method `return_book()` that sets `available` back to `True` and prints a confirmation message. (2 marks)
4. Has a `__str__` method that returns a readable summary of the book, for example: `"Frankenstein by Mary Shelley (1818) - Available"`. (2 marks)

Create **two** `Book` objects, test all methods, and print the results.

> **Hint:** Use `self` to refer to the object's own attributes inside the class. The `__init__` method runs automatically when you create an object: `book1 = Book("Title", "Author", 2001)`. The `__str__` method is called automatically when you `print()` an object.

````python
from IPython.terminal.interactiveshell import TerminalInteractiveShell
# Answer to Question 3

class Book():
  def __init__ (self, title, author, year, available = True):
    self.title = title
    self.author = author
    self.year = year
    self.available = available

  def checkout(self):
    if self.available == True:
      self.available = False
      print(f"you have successfully checked out the book: {self.title}.")
    else:
      print(f"Sorry! the book: {self.title} is not available at the moment.")

  def return_book(self):
    if self.available == False:
      self.available = True
      print(f"you have successfully returned book: {self.title}.")
    else:
      print(f"The book: {self.title} has already been returned.")

  def __str__(self):
     return (f"Summary of the book: Name: {self.title}. Author: {self.author}. Released in: {self.year}. Availability: {self.available}")


book1 = Book("Ohayo Sekai", "Senku","2022")
book2 = Book("Spider", "Kafka","2015")

book1.checkout()
book1.checkout()
book1.return_book()
book1.return_book()
book1.__str__()

print("\n\n")
book2.checkout()
book2.checkout()
book2.return_book()
book2.return_book()
book2.__str__()

````
